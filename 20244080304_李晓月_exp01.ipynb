```python
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("成绩超出范围")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"


print(score_level(95))
print(score_level(85))
print(score_level(70))
print(score_level(50))

try:
    print(score_level(120))
except ValueError as e:
    print(e)


sum_for = 0
for i in range(1, 101):
    sum_for += i

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1

print("for结果:", sum_for)
print("while结果:", sum_while)
print("结果是否相同:", sum_for == sum_while)


def factorial_recursive(n):
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


for n in [0, 1, 5, 10]:
    print(n, factorial_recursive(n), factorial_iterative(n))


def summary(values):
    return min(values), max(values), sum(values) / len(values)


values = [10, 20, 30, 40, 50]

min_value, max_value, avg_value = summary(values)

print("最小值:", min_value)
print("最大值:", max_value)
print("平均值:", avg_value)


def sum_all(*args):
    return sum(args)


print(sum_all())
print(sum_all(10))
print(sum_all(1, 2, 3, 4, 5))
```

```python
import os
import pickle


work_dir = os.getcwd()
data_dir = os.path.join(work_dir, "exp01_data")

os.makedirs(data_dir, exist_ok=True)

file_path = os.path.join(data_dir, "students.txt")


data = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete"""

with open(file_path, "w", encoding="utf-8") as f:
    f.write(data)


students = []

with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

for line in lines[1:]:
    print(line.strip())


def parse_student(line):
    parts = line.strip().split(",")

    if len(parts) != 2:
        raise ValueError("字段数量错误")

    name = parts[0]

    if name == "":
        raise ValueError("姓名为空")

    score = int(parts[1])

    if score < 0 or score > 100:
        raise ValueError("成绩超出范围")

    return {
        "name": name,
        "score": score
    }


valid_students = []
invalid_records = []

with open(file_path, "r", encoding="utf-8") as f:
    next(f)

    for line_number, line in enumerate(f, start=2):
        try:
            student = parse_student(line)
            valid_students.append(student)
        except ValueError as e:
            invalid_records.append((line_number, str(e)))


print("有效记录:", valid_students)
print("无效记录:", invalid_records)


count = len(valid_students)
average = sum(s["score"] for s in valid_students) / count
pass_rate = sum(1 for s in valid_students if s["score"] >= 60) / count

report_path = os.path.join(data_dir, "report.txt")

with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {count}\n")
    f.write(f"无效记录数: {len(invalid_records)}\n")
    f.write(f"平均分: {average:.2f}\n")
    f.write(f"及格率: {pass_rate:.2%}\n")


with open(report_path, "a", encoding="utf-8") as f:
    f.write("追加内容测试\n")


pickle_path = os.path.join(data_dir, "students.pkl")

with open(pickle_path, "wb") as f:
    pickle.dump(valid_students, f)


with open(pickle_path, "rb") as f:
    loaded_students = pickle.load(f)

print(valid_students == loaded_students)


with open("student_utils.py", "w", encoding="utf-8") as f:
    f.write(
"""def parse_student(line):
    parts = line.strip().split(",")

    if len(parts) != 2:
        raise ValueError("字段数量错误")

    name = parts[0]

    if name == "":
        raise ValueError("姓名为空")

    score = int(parts[1])

    if score < 0 or score > 100:
        raise ValueError("成绩超出范围")

    return {
        "name": name,
        "score": score
    }
"""
    )


from student_utils import parse_student

print(parse_student("Tom,90"))

try:
    print(parse_student("Tom,abc"))
except ValueError as e:
    print(e)


try:
    with open("not_exist.txt", "r", encoding="utf-8") as f:
        f.read()
except FileNotFoundError:
    print("文件不存在")
```

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"姓名:{self.name}, 成绩:{self.score}"


valid_students = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 58},
    {"name": "Carol", "score": 92},
    {"name": "David", "score": 76},
    {"name": "Eva", "score": 60},
    {"name": "Frank", "score": 89}
]


students = []

for s in valid_students:
    student = Student(s["name"], s["score"])
    students.append(student)
    print(student)
    print("是否及格:", student.is_passed())


class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"姓名:{self.name}, 成绩:{self.score}, 研究方向:{self.research_topic}"


graduate = GraduateStudent("Tom", 90, "Data Analysis")

print(graduate)


import copy


obj1 = {"name": "Alice", "scores": [85, 90]}

obj2 = obj1
obj2["scores"].append(95)

print("直接赋值:")
print("obj1:", obj1)
print("obj2:", obj2)


obj1 = {"name": "Alice", "scores": [85, 90]}

obj3 = copy.copy(obj1)
obj3["scores"].append(100)

print("浅复制:")
print("obj1:", obj1)
print("obj3:", obj3)


obj1 = {"name": "Alice", "scores": [85, 90]}

obj4 = copy.deepcopy(obj1)
obj4["scores"].append(100)

print("深复制:")
print("obj1:", obj1)
print("obj4:", obj4)
```
