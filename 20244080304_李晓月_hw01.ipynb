```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m


pairs = [
    (2018, 2019),
    (1200, 300),
    (5040, 60)
]

for m, n in pairs:
    print(f"gcd({m}, {n}) =", gcd(m, n))


print(gcd(-12, 8))
print(gcd(12, -8))
print(gcd(-12, -8))
```

```python
import math


def euler_limit(n):
    return (1 + 1 / n) ** n


def euler_infinite_sum(n):
    if n == 0:
        return 0

    result = 0
    factorial = 1

    for k in range(n):
        if k > 0:
            factorial *= k
        result += 1 / factorial

    return result


def euler_approx(epsilon):
    n = 1

    while abs(euler_infinite_sum(n) - math.exp(1)) >= epsilon:
        n += 1

    return euler_infinite_sum(n)


def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))


print(euler_limit(10000))

print(euler_infinite_sum(4))
print(euler_infinite_sum(1))
print(euler_infinite_sum(0))

print(euler_approx(0.00001))

print_euler_sum_table(10)
```

```python
def is_power_of_2(n):
    while n % 2 == 0:
        n = n // 2

    return n == 1


def is_power(b, n):
    while n % b == 0:
        n = n // b

    return n == 1


print(is_power_of_2(1))
print(is_power_of_2(2))
print(is_power_of_2(4))
print(is_power_of_2(8))
print(is_power_of_2(10))
print(is_power_of_2(32))


print(is_power(2, 16))
print(is_power(3, 27))
print(is_power(5, 125))
print(is_power(2, 18))
```
