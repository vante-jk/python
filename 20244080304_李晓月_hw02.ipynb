```python
# ==================== 字符串的乐趣 ====================

def is_palindrome(s):
    """忽略空格和大小写，判断字符串是否为回文。"""
    s = s.replace(" ", "").lower()
    return s == s[::-1]


def is_abecedarian(s):
    """忽略空格，判断字母是否按字母顺序排列。"""
    s = s.replace(" ", "").lower()
    return all(s[i] <= s[i + 1] for i in range(len(s) - 1))


def remove_vowels(s):
    """删除字符串中的所有元音字母。"""
    vowels = "aeiouAEIOU"
    return "".join(char for char in s if char not in vowels)


# 测试
print("【回文判断】")
print(is_palindrome("level"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("Was it a car or a cat I saw?"))
print(is_palindrome("hello"))

print("\n【字母顺序判断】")
print(is_abecedarian("adder"))
print(is_abecedarian("beet"))
print(is_abecedarian("dog"))
print(is_abecedarian("abcd efgh xyz"))

print("\n【删除元音】")
print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("AEIOU"))
print(remove_vowels("Hello World"))
```

```python
# ==================== 列表的乐趣 ====================

def list_reverse(lst):
    """返回反转后的列表；输入不是列表时引发 TypeError。"""
    if not isinstance(lst, list):
        raise TypeError("输入必须是一个列表")
    return lst[::-1]


def is_sorted(seq):
    """判断序列是否按升序排列，只遍历序列一次。"""
    for i in range(1, len(seq)):
        if seq[i - 1] > seq[i]:
            return False
    return True


# 测试 list_reverse
print("【列表反转】")
print(list_reverse([1, 2, 3]))
print(list_reverse(["a", "b", "c"]))
print(list_reverse([]))

# 测试非列表输入
try:
    list_reverse((1, 2, 3))
except TypeError as e:
    print("异常测试：", e)

# 测试 is_sorted
print("\n【升序判断】")
print(is_sorted([1, 2, 3, 4]))
print(is_sorted([1, 1, 2, 3]))
print(is_sorted([3, 2, 1]))
print(is_sorted([]))
print(is_sorted((1, 2, 3)))
```

```python
# ==================== 作为向量的元组 ====================

def vec_scalar_mult(t, s):
    """计算元组向量与标量的乘积，返回元组。"""
    if not isinstance(t, tuple):
        raise TypeError("向量必须是元组")
    if not all(isinstance(x, (int, float)) for x in t):
        raise TypeError("向量元素必须是整数或浮点数")
    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或浮点数")

    return tuple(x * s for x in t)


def vec_inner_product(t1, t2):
    """计算两个元组向量的内积，返回浮点数。"""
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个向量都必须是元组")

    if len(t1) != len(t2):
        raise ValueError("两个向量的长度必须相同")

    if not all(isinstance(x, (int, float)) for x in t1 + t2):
        raise TypeError("向量元素必须是整数或浮点数")

    return float(sum(x * y for x, y in zip(t1, t2)))


def check_valid_mx(my_mx):
    """检查参数是否为有效矩阵：元组的元组，且每个元素均为数值。"""
    if not isinstance(my_mx, tuple):
        return False

    if len(my_mx) == 0:
        return False

    if not all(isinstance(row, tuple) for row in my_mx):
        return False

    row_length = len(my_mx[0])

    if not all(len(row) == row_length for row in my_mx):
        return False

    if not all(
        isinstance(value, (int, float))
        for row in my_mx
        for value in row
    ):
        return False

    return True


def mx_vec_mult(my_mx, vec):
    """计算矩阵与向量的乘积，返回数值元组。"""
    if not check_valid_mx(my_mx):
        raise ValueError("输入矩阵无效")

    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")

    if not all(isinstance(x, (int, float)) for x in vec):
        raise TypeError("向量元素必须是整数或浮点数")

    if len(my_mx[0]) != len(vec):
        raise ValueError("矩阵的列数必须等于向量的长度")

    return tuple(
        float(sum(a * b for a, b in zip(row, vec)))
        for row in my_mx
    )


# 测试任务1：向量与标量相乘
print("【向量与标量相乘】")
print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.5, 2, 3.5), 2.0))
print(vec_scalar_mult((1, 2, 3), 0))

# 测试任务2：向量内积
print("\n【向量内积】")
print(vec_inner_product((1, 2, 3), (4, 5, 6)))
print(vec_inner_product((1.5, 2.0), (2, 3)))

# 测试内积异常
try:
    vec_inner_product((1, 2), (3, 4, 5))
except ValueError as e:
    print("异常测试：", e)

# 测试任务3：矩阵合法性
print("\n【矩阵合法性检查】")
my_mx = ((1, 2, 3), (4, 5, 6))
print("有效矩阵：", check_valid_mx(my_mx))
print("内层不是元组：", check_valid_mx(([1, 2], [3, 4])))
print("行长度不一致：", check_valid_mx(((1, 2), (3, 4, 5))))
print("含有非数值元素：", check_valid_mx(((1, 2), (3, "a"))))

# 测试任务4：矩阵与向量相乘
print("\n【矩阵与向量相乘】")
print(mx_vec_mult(((1, 2, 3), (4, 5, 6)), (1, 2, 3)))

# 测试矩阵与向量维度不匹配
try:
    mx_vec_mult(((1, 2), (3, 4)), (1, 2, 3))
except ValueError as e:
    print("异常测试：", e)
```

```python
# ==================== 更多向量的乐趣 ====================

class Vector:
    """使用列表或元组表示元素的向量。"""

    def __init__(self, dimension, elements=None):
        # 检查维度
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("维度必须是非负整数")

        if dimension < 0:
            raise ValueError("维度不能为负数")

        # 未提供元素时，默认创建零向量
        if elements is None:
            elements = (0,) * dimension

        # 检查元素容器类型
        if not isinstance(elements, (list, tuple)):
            raise TypeError("向量元素必须使用列表或元组提供")

        # 检查元素类型
        if not all(isinstance(x, (int, float)) for x in elements):
            raise TypeError("向量元素必须是整数或浮点数")

        # 检查维度与元素数量是否一致
        if len(elements) != dimension:
            raise ValueError("维度与元素数量不一致")

        self.dimension = dimension
        self.elements = tuple(elements)

    def _check_same_dimension(self, other):
        """检查另一个向量的类型和维度。"""
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector 对象")

        if self.dimension != other.dimension:
            raise ValueError("两个向量的维度必须相同")

    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._check_same_dimension(other)
        return self.elements == other.elements

    def __lt__(self, other):
        self._check_same_dimension(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check_same_dimension(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_same_dimension(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_same_dimension(other)
        return self.elements >= other.elements

    def dot(self, other):
        """计算向量内积。"""
        self._check_same_dimension(other)
        return float(sum(
            x * y for x, y in zip(self.elements, other.elements)
        ))

    def __mul__(self, other):
        """支持标量乘法和两个向量的逐元素乘法。"""
        if isinstance(other, Vector):
            self._check_same_dimension(other)
            result = tuple(
                x * y for x, y in zip(self.elements, other.elements)
            )
            return Vector(self.dimension, result)

        if isinstance(other, (int, float)):
            result = tuple(x * other for x in self.elements)
            return Vector(self.dimension, result)

        return NotImplemented

    def __rmul__(self, other):
        """支持标量左乘，例如 2 * v。"""
        return self.__mul__(other)


# 任务1：创建向量对象
print("【任务1：创建向量】")
v1 = Vector(3, (1, 2, 3))
v2 = Vector(3, [1, 2, 3])
v3 = Vector(3)

print("v1 =", v1)
print("v2 =", v2)
print("默认零向量 =", v3)

# 任务2：维度和元素应当是实例属性
print("\n【任务2：属性设计】")
print("dimension 和 elements 是实例属性。")
print("理由：不同 Vector 对象可以具有不同维度和元素，")
print("因此这些数据应属于具体对象，而不是由所有对象共享的类属性。")

# 任务3：比较运算符
print("\n【任务3：向量比较】")
a = Vector(2, (2, 4))
b = Vector(2, (2, 5))
c = Vector(2, (2, 4))

print("a == c：", a == c)
print("a < b：", a < b)
print("a <= c：", a <= c)
print("b > a：", b > a)
print("b >= a：", b >= a)

# 测试维度不一致的比较
try:
    print(a < Vector(3, (1, 2, 3)))
except ValueError as e:
    print("异常测试：", e)

# 任务4：向量内积
print("\n【任务4：Vector.dot】")
v4 = Vector(3, (1, 2, 3))
v5 = Vector(3, (4, 5, 6))
print("内积：", v4.dot(v5))

# 测试 dot 的维度异常
try:
    v4.dot(Vector(2, (1, 2)))
except ValueError as e:
    print("异常测试：", e)

# 任务5：标量乘法和向量逐元素乘法
print("\n【任务5：Vector 乘法运算符】")
v6 = Vector(2, (2, 3))
v7 = Vector(2, (4, 5))

print("2 * v6 =", 2 * v6)
print("v6 * 2 =", v6 * 2)
print("v6 * v7 =", v6 * v7)

# 测试不同维度向量相乘
try:
    print(v6 * Vector(3, (1, 2, 3)))
except ValueError as e:
    print("异常测试：", e)

# 测试非法维度
try:
    Vector(-1)
except ValueError as e:
    print("异常测试：", e)

try:
    Vector(2, (1, 2, 3))
except ValueError as e:
    print("异常测试：", e)
```
