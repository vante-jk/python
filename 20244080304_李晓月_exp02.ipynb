```python
# ==================== 实验一：NumPy 数组基础与运算 ====================

# 任务1：创建不同类型的 ndarray，查看 dtype、shape 和 ndim
arr_int = np.array([1, 2, 3, 4])
arr_float = np.array([1.2, 2.5, 3.8])
arr_bool = np.array([True, False, True])

print("【任务1：数组类型与属性】")
for name, arr in [
    ("整数数组", arr_int),
    ("浮点数组", arr_float),
    ("布尔数组", arr_bool)
]:
    print(f"{name}：{arr}")
    print(f"dtype={arr.dtype}, shape={arr.shape}, ndim={arr.ndim}")

print("\nNumPy 数组与 Python 列表的区别：")
print("NumPy 数组通常适合存储同类型数据，支持向量化运算和多维数组操作；")
print("Python 列表可以存储不同类型的对象，但数值运算通常需要循环或其他操作。")


# 任务2：使用 arange 和 reshape 创建多维数组
arr_12 = np.arange(12)
arr_3x4 = arr_12.reshape(3, 4)
arr_2x3x2 = arr_12.reshape(2, 3, 2)

print("\n【任务2：数组变形】")
print("原始数组：\n", arr_12)
print("变形为 (3, 4)：\n", arr_3x4)
print("变形为 (2, 3, 2)：\n", arr_2x3x2)
print("三维数组的 shape：", arr_2x3x2.shape)


# 任务3：数组索引、切片和布尔索引
arr = np.arange(1, 13).reshape(3, 4)

second_row = arr[1, :]
selected_columns = arr[:, 0:3]
greater_than_6 = arr[arr > 6]

print("\n【任务3：索引、切片与布尔索引】")
print("原数组：\n", arr)
print("第2行：", second_row, "shape =", second_row.shape)
print("所有行的第1至第3列：\n", selected_columns)
print("大于6的元素：", greater_than_6)
print("大于6的元素 shape：", greater_than_6.shape)


# 任务4：使用逻辑条件筛选数组
arr_10 = np.arange(10)

condition_or = (arr_10 < 2) | (arr_10 > 7)
condition_and = (arr_10 > 3) & (arr_10 < 7)

result_or = arr_10[condition_or]
result_and = arr_10[condition_and]

print("\n【任务4：逻辑条件筛选】")
print("原数组：", arr_10)
print("小于2或大于7：", result_or)
print("大于3且小于7：", result_and)


# 任务5：比较逐元素乘法和矩阵乘法
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("\n【任务5：逐元素乘法与矩阵乘法】")
print("矩阵 A：\n", A)
print("矩阵 B：\n", B)
print("A * B：\n", A * B)
print("np.multiply(A, B)：\n", np.multiply(A, B))
print("A @ B：\n", A @ B)
print("np.matmul(A, B)：\n", np.matmul(A, B))

print("说明：* 和 np.multiply() 执行对应元素相乘；")
print("@ 和 np.matmul() 执行矩阵乘法。")
```

```python
# ==================== 实验二：随机数、统计与缺失值处理 ====================

# 任务1：生成1000个模拟成绩，并限制在0至100分
rng = np.random.default_rng(2026)

scores = rng.normal(loc=70, scale=10, size=1000)
scores = np.clip(scores, 0, 100)

print("【任务1：模拟成绩统计】")
print("样本数量：", len(scores))
print("最小值：", np.min(scores))
print("最大值：", np.max(scores))
print("平均值：", np.mean(scores))
print("标准差：", np.std(scores))
print("极差：", np.ptp(scores))


# 任务2：不放回均匀抽样和按整数大小加权抽样
numbers = np.arange(1, 11)

uniform_sample = rng.choice(numbers, size=5, replace=False)

weights = numbers / numbers.sum()
weighted_sample = rng.choice(
    numbers, size=5, replace=False, p=weights
)

print("\n【任务2：不放回抽样】")
print("原始整数：", numbers)
print("均匀抽样结果：", uniform_sample)
print("按整数大小加权抽样结果：", weighted_sample)
print("加权抽样中，数值越大的整数被抽中的概率越高。")


# 任务3：比较 permutation 和 shuffle
original = np.arange(1, 11)

permutation_result = rng.permutation(original)

print("\n【任务3：permutation 与 shuffle】")
print("permutation 前的原数组：", original)
print("permutation 的结果：", permutation_result)
print("permutation 后的原数组：", original)

shuffle_array = original.copy()
print("shuffle 前的数组：", shuffle_array)
rng.shuffle(shuffle_array)
print("shuffle 后的数组：", shuffle_array)
print("说明：permutation 返回打乱后的新数组；shuffle 原地修改数组。")


# 任务4：设置缺失值，比较均值计算结果
scores_with_nan = scores.copy()
scores_with_nan[[9, 19, 29]] = np.nan

ordinary_mean = np.mean(scores_with_nan)
nan_mean = np.nanmean(scores_with_nan)
nan_count = np.isnan(scores_with_nan).sum()

print("\n【任务4：缺失值处理】")
print("缺失值数量：", nan_count)
print("包含 NaN 时的普通均值：", ordinary_mean)
print("忽略 NaN 后的均值：", nan_mean)
print("说明：普通 mean 遇到 NaN 通常会返回 NaN；")
print("nanmean 会忽略缺失值，计算有效数据的平均值。")


# 任务5：建立内存中的成绩表，供实验三使用
scores_df = pd.DataFrame({
    "Student_No": np.arange(1, len(scores_with_nan) + 1),
    "Score": scores_with_nan
})

print("\n【任务5：建立内存成绩表】")
print(scores_df.head(12))
print("成绩表形状：", scores_df.shape)
print("说明：数据保存在内存变量中，不写入磁盘。")
```

```python
# ==================== 实验三：SciPy 科学计算与 Matplotlib 可视化 ====================

from scipy import linalg, optimize, stats

# 任务1：使用 scipy.linalg.solve 求解线性方程组
# 3x + y = 9
# x + 2y = 8
A = np.array([[3, 1], [1, 2]], dtype=float)
b = np.array([9, 8], dtype=float)

solution = linalg.solve(A, b)
x_value, y_value = solution

print("【任务1：线性方程组求解】")
print("求解结果：", solution)
print("代入验证：", A @ solution)
print("原方程右侧：", b)
print("验证是否成立：", np.allclose(A @ solution, b))


# 任务2：使用 scipy.optimize.root 求 x² - 2 = 0 的正根和负根
def equation(x):
    return x**2 - 2

positive_root = optimize.root(equation, x0=1.0)
negative_root = optimize.root(equation, x0=-1.0)

print("\n【任务2：非线性方程求根】")
print("正根求解 success：", positive_root.success)
print("正根：", positive_root.x[0])
print("正根对应的函数值：", equation(positive_root.x[0]))

print("负根求解 success：", negative_root.success)
print("负根：", negative_root.x[0])
print("负根对应的函数值：", equation(negative_root.x[0]))


# 任务3：使用 minimize 最小化 x² + x + 2
def objective(x):
    return x[0]**2 + x[0] + 2

minimum_result = optimize.minimize(objective, x0=[0.0])

print("\n【任务3：函数最小值】")
print("求解 success：", minimum_result.success)
print("数值计算得到的极小点：", minimum_result.x[0])
print("数值计算得到的最小值：", minimum_result.fun)

# 解析结果：x² + x + 2 = (x + 0.5)² + 1.75
print("解析极小点：", -0.5)
print("解析最小值：", 1.75)


# 任务4：两组独立样本 t 检验
rng_test = np.random.default_rng(2026)
sample_1 = rng_test.normal(loc=0, scale=1, size=100)
sample_2 = rng_test.normal(loc=1, scale=1, size=100)

t_result = stats.ttest_ind(sample_1, sample_2)

print("\n【任务4：独立样本 t 检验】")
print("第一组样本均值：", np.mean(sample_1))
print("第二组样本均值：", np.mean(sample_2))
print("t 统计量：", t_result.statistic)
print("p 值：", t_result.pvalue)

if t_result.pvalue < 0.05:
    print("结论：p < 0.05，拒绝两组总体均值相等的原假设。")
else:
    print("结论：p >= 0.05，不能拒绝两组总体均值相等的原假设。")


# 任务5：使用实验二成绩绘制直方图和箱线图
valid_scores = scores_df["Score"].dropna().to_numpy()

plt.figure(figsize=(8, 5))
plt.hist(valid_scores, bins=20, edgecolor="black")
plt.title("Distribution of Simulated Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
plt.boxplot(valid_scores)
plt.title("Boxplot of Simulated Scores")
plt.ylabel("Score")
plt.tight_layout()
plt.show()


# 任务6：在同一坐标系绘制 y=x、y=x²、y=x³
x = np.linspace(0, 5, 200)

plt.figure(figsize=(9, 6))
plt.plot(x, x, label="y = x")
plt.plot(x, x**2, label="y = x^2")
plt.plot(x, x**3, label="y = x^3")

plt.title("Functions: y=x, y=x^2, y=x^3")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# ==================== 实验四：pandas 数据处理与分析 ====================

from io import StringIO

# 任务1：创建指定学生成绩 DataFrame，并以姓名作为行索引
records = {
    "Name": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "Group": ["A", "A", "B", "B", "A", "B"],
    "Math": [88, 72, 91, 65, 79, 84],
    "English": [90, 68, 85, 73, 92, 77],
    "Python": [95, 80, 89, 70, 86, 82]
}

df = pd.DataFrame(records).set_index("Name")

# 计算总分和平均分
score_columns = ["Math", "English", "Python"]
df["Total"] = df[score_columns].sum(axis=1)
df["Average"] = df[score_columns].mean(axis=1)

print("【任务1：创建 DataFrame 和基本统计】")
print(df)
print("\n行索引：", df.index)
print("列名：", df.columns)
print("\n前3行：")
print(df.head(3))
print("\n描述性统计：")
print(df.describe())


# 任务2：使用 loc、iloc 和条件筛选
print("\n【任务2：索引与条件筛选】")
print("使用 loc 读取 Carol：")
print(df.loc["Carol"])

print("\n使用 iloc 读取第3行（Carol）：")
print(df.iloc[2])

print("\n筛选平均分大于等于80的学生：")
print(df[df["Average"] >= 80])


# 任务3：axis=0 和 axis=1 的区别
print("\n【任务3：按列和按行求均值】")
print("三科各自的平均分：")
print(df[score_columns].mean(axis=0))

print("\n每位学生三科的平均分：")
print(df[score_columns].mean(axis=1))


# 任务4：使用 apply、agg 和 rank
print("\n【任务4：apply、agg 和 rank】")
df["Highest_Score"] = df[score_columns].apply(max, axis=1)
print("每位学生的三科最高分：")
print(df[["Highest_Score"]])

print("\n每科最低分、平均分、最高分：")
print(df[score_columns].agg(["min", "mean", "max"]))

print("\n数学成绩排名（1表示最高分）：")
print(df["Math"].rank(ascending=False, method="min"))


# 任务5：按组统计人数和三科平均分
print("\n【任务5：分组统计】")
group_summary = df.groupby("Group").agg(
    Student_Count=("Group", "size"),
    Math_Mean=("Math", "mean"),
    English_Mean=("English", "mean"),
    Python_Mean=("Python", "mean")
)
print(group_summary)


# 任务6：使用 transform 计算所在组的 Python 平均分
df["Group_Python_Mean"] = df.groupby("Group")["Python"].transform("mean")

print("\n【任务6：transform】")
print(df[["Group", "Python", "Group_Python_Mean"]])


# 任务7：使用 filter 保留 Python 平均分不低于80的组
filtered_groups = df.groupby("Group").filter(
    lambda group: group["Python"].mean() >= 80
)

print("\n【任务7：filter】")
print("保留 Python 平均分 >= 80 的组：")
print(filtered_groups[["Group", "Python"]])
print("保留的组：", filtered_groups["Group"].unique())


# 任务8：协方差矩阵和 Pearson 相关矩阵
print("\n【任务8：协方差和相关性】")
cov_matrix = df[score_columns].cov()
corr_matrix = df[score_columns].corr(method="pearson")

print("三科协方差矩阵：")
print(cov_matrix)

print("\n三科 Pearson 相关系数矩阵：")
print(corr_matrix)

print("\n解释示例：")
print(
    "协方差为正，表示两科成绩总体上倾向于同方向变化；"
    "协方差为负，表示倾向于反方向变化。"
)
print(
    "Pearson 相关系数范围为 -1 到 1，绝对值越接近1，"
    "线性相关程度通常越强；接近0表示线性相关较弱。"
)


# 任务9：在内存中演示 CSV 读写，不创建磁盘文件
print("\n【任务9：内存中的 CSV 读写】")
csv_buffer = StringIO()
df.to_csv(csv_buffer, index=True)

csv_text = csv_buffer.getvalue()
print("生成的 CSV 文本前几行：")
print("\n".join(csv_text.splitlines()[:4]))

csv_buffer.seek(0)
df_from_csv = pd.read_csv(csv_buffer, index_col=0)

print("\n从内存 CSV 读取后的数据：")
print(df_from_csv.head())

print("\n说明：使用 StringIO 在内存中完成 CSV 转换，未在磁盘上创建文件。")
```
